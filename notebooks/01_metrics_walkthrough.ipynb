## Sequence distance and Unicode normalization
Hebrew Unicode is stored in logical order. We use NFD by default and keep combining marks unless explicitly changed.


```python
from unicodedata import normalize

def edit_distance(a, b):
    previous = list(range(len(b) + 1))
    for i, x in enumerate(a, 1):
        current = [i]
        for j, y in enumerate(b, 1):
            current.append(min(current[-1] + 1, previous[j] + 1, previous[j-1] + (x != y)))
        previous = current
    return previous[-1]
```

```python
synthetic_reference = normalize("NFD", "אבג דה")
synthetic_prediction = normalize("NFD", "אב דה")
cer = edit_distance(synthetic_reference, synthetic_prediction) / len(synthetic_reference)
wer = edit_distance(synthetic_reference.split(), synthetic_prediction.split()) / len(synthetic_reference.split())
print({"CER": cer, "WER": wer})
```
